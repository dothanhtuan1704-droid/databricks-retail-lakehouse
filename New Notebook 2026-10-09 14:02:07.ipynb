```python
# Notebook kiểm định chất lượng dữ liệu tầng Bronze
tables = ["bronze_stores", "bronze_products", "bronze_customers", "bronze_sales", "bronze_exchange_rates"]

print(" BẮT ĐẦU KIỂM TRA CHẤT LƯỢNG DỮ LIỆU TẦNG BRONZE...")

for tbl in tables:
    count = spark.table(f"workspace.bronze.{tbl}").count()
    # Kiểm tra xem bảng có bị rỗng không (Data Quality Check)
    assert count > 0, f"LỖI NGHIÊM TRỌNG: Bảng {tbl} không có dữ liệu!"
    print(f" {tbl}: {count:,} dòng -> ĐẠT TIÊU CHUẨN")

print("\n TẤT CẢ DỮ LIỆU BRONZE ĐÃ SẴN SÀNG CHO TẦNG SILVER!")
```
